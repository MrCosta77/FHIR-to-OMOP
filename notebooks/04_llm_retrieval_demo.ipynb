```python
import ollama

# The exact model you installed on your machine
MODEL_NAME = "qwen2.5-coder:7b"

# A sample of unstructured free-text written by a doctor
clinical_note = """
Patient presents with severe abdominal pain and nausea. 
He mentions a history of Type 2 Diabetes Mellitus.
Suspecting acute appendicitis.
"""

# The system prompt dictates the rules of engagement for the LLM
system_prompt = """
You are an expert Clinical Data Informatician. 
Your task is to read clinical notes and extract only the medical conditions mentioned.
Return the conditions as a simple comma-separated list. Do not add conversational text.
"""

print(f"🔌 Connecting to local Ollama server (Model: {MODEL_NAME})...")
print("🧠 Asking Qwen to analyze the clinical note...\n")

try:
    # The local API call (completely offline and private)
    response = ollama.chat(
        model=MODEL_NAME,
        messages=[
            {'role': 'system', 'content': system_prompt},
            {'role': 'user', 'content': f"Clinical Note:\n{clinical_note}"}
        ]
    )
    
    # Print the final output from the model
    print("-" * 50)
    print("📝 EXTRACTED RESULTS BY QWEN:")
    print(response['message']['content'])
    print("-" * 50)
    
    print("\n✅ Success! Your Python environment and the local LLM are communicating perfectly.")

except Exception as e:
    print(f"\n❌ Connection Error: {e}")
    print("Please verify if the Ollama app is running in Windows and the model is downloaded.")
```
