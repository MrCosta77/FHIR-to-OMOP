```python
import json
import os

# Setup paths (relative to the 'omop-etl' directory)
FHIR_DIR = "../synthea/output/fhir"

def explore_patient_bundle(patient_file):
    """
    Reads a FHIR bundle JSON and extracts high-level patient insights.
    """
    file_path = os.path.join(FHIR_DIR, patient_file)

    with open(file_path, encoding='utf-8') as f:
        fhir_data = json.load(f)

    entries = fhir_data.get('entry', [])
    print(f"✅ Processing patient: {patient_file}")
    print(f"Total FHIR entries (events) found: {len(entries)}")

    # Extract clinical conditions
    conditions = []
    for entry in entries:
        resource = entry.get('resource', {})
        if resource.get('resourceType') == 'Condition':
            condition_name = resource.get('code', {}).get('text', 'Unknown')
            recorded_date = resource.get('recordedDate', 'No Date')[:10]
            conditions.append({'Date': recorded_date, 'Condition': condition_name})

    print(f"\nExtracted {len(conditions)} clinical conditions. Top 5 most recent:")
    for c in reversed(conditions[-5:]):
        print(f" -> {c['Date']}: {c['Condition']}")

# Execution block
json_files = [f for f in os.listdir(FHIR_DIR) if f.endswith('.json')]
if json_files:
    # Pick the first patient as a sample
    sample_patient = json_files[0]
    explore_patient_bundle(sample_patient)
else:
    print("❌ No JSON files found. Check if Synthea data generation was successful.")
```
